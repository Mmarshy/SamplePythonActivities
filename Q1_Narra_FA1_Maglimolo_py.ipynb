```python
import math

# Get the coordinates from the user
x1 = float(input(str("What is your X1?: ")))
x2 = float(input(str("What is your X2?: ")))
y1 = float(input(str("What is your Y1?: ")))
y2 = float(input(str("What is your Y2?: ")))

# Calculate the distance between the two points
distance = math.sqrt(pow(x2 - x1, 2) + pow(y2 - y1, 2))

# Round the distance to 2 decimal places
rounded_distance = round(distance, 2)

# Display the result
print("The distance between your coordinates is:")
print(rounded_distance)

#Using a library is more practical because it provides ready made functions, saving time and reducing the chance of errors. In my activity, I used the math library's sqrt() function to calculate the square root needed for the distance formula instead of creating my own square-root calculation.
```
