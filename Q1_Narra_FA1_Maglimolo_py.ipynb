```python
import math
X1 =int(input(str("what is your X1?: ")))
X2 =int(input(str("what is your X2?: ")))
Y1 =int(input(str("what is your Y1?: ")))
Y2 =int(input(str("what is your Y2?: ")))

#finds the distance between the two points
distance = math.sqrt(math.pow(X2-X1 , 2) + math.pow(Y2-Y1 , 2))

#this rounds the number to 2 decimal points
rounded_distance = round(distance,2)

print("the distance between your coordinates is")
print(rounded_distance)

#it is mopre practical in using libraries than making it from scratch because libraries  have the code and is just waiting for you to use it but if you make it from scratch it take you longer be cause you still have to make it than use an existing code
```
