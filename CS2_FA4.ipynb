```python
import re
from datetime import datetime

# Get the current year
current_year = datetime.now().year

# AGE VALIDATION
while True:
    try:
        age_input = input("Enter your age: ")

        if age_input.strip() == "":
            print("Invalid input! You did not enter anything.")
            continue

        age = int(age_input)

        if age < 12:
            print("Invalid age! You must be at least 12 years old.")
        elif age > 17:
            print("Invalid age! You must be 17 years old or younger.")
        else:
            print("Age accepted:", age)
            break

    except ValueError:
        print("Invalid input! Please enter a whole number.")


# GRADE LEVEL VALIDATION
while True:
    try:
        grade_input = input("Enter your grade level: ")

        if grade_input.strip() == "":
            print("Invalid input! You did not enter anything.")
            continue

        grade = int(grade_input)

        if 7 <= grade <= 10:
            print("Grade level accepted:", grade)
            break
        else:
            print("Invalid grade level! Please enter a grade from 7 to 10.")

    except ValueError:
        print("Invalid input! Please enter a whole number.")


# PARTICIPANT CODE VALIDATION
while True:
    code = input("Enter your participant code: ")

    if code.strip() == "":
        print("Invalid input! You did not enter anything.")
        continue

    pattern = rf"CS-2-\d{{4}}-{current_year}"

    if re.fullmatch(pattern, code):
        print("Participant code accepted:", code)
        break
    else:
        print(f"Invalid participant code! Please use the format CS-2-XXXX-{current_year}.")


print("\nAll information is valid!")
print("Age:", age)
print("Grade Level:", grade)
print("Participant Code:", code)
```
