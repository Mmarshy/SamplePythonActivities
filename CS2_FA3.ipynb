```python
# Ask for credit score and check it right away to fit the flowchart
credit_score = float(input("Enter credit score: "))

if credit_score < 700:
    print("Result: Loan Denied")
else:
    # Only reached if credit score passed, Now ask the user for income
    annual_income = float(input("Enter annual income: "))

    if annual_income < 30000:
        print("Result: Loan Denied")
    else:
        # Only reached if credit score AND income passed Now ask the user for job stability
        years_at_job = float(input("Enter years at current job: "))

        if years_at_job < 2:
            print("Result: Loan Denied")
        else:
            # Only reached if all three checks passed.
            print("Result: Loan Approved")
```
